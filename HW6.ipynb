```python
import os
path = '/content/drive/My Drive/PythonDataViz2026'
file_path = os.path.join(path,'confirmed-cases.csv')
print(file_path)
```

```python
data_covid = pd.read_csv(file_path)
data_covid.head()
```

```python
data_covid.shape
```

```python
data = data_covid[['sex','age','province_of_onset']]
data
```

**1.   คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของข้อมูลทั้งหมด**

```python
female = data[data['sex'] == 'หญิง']
female_numeric_age = pd.to_numeric(female['age'], errors='coerce')
print(female_numeric_age.head())
```

```python
sum = 0
N = 0
for a in female_numeric_age:
    if a > 0:
        sum += a
        N += 1
print(f'อายุเฉลี่ยของผู้ป่วยหญิง:{sum/N}')
```

```python
male = data[data['sex'] == 'ชาย']
male_numeric_age = pd.to_numeric(male['age'], errors='coerce')
print(male_numeric_age.head())
```

```python
sum = 0
N = 0
for a in male_numeric_age:
    if a > 0:
        sum += a
        N += 1
print(f'อายุเฉลี่ยของผู้ป่วยชาย:{sum/N}')
```

**2.   คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของผู้ป่วยในจังหวัดขอนแก่น**

```python
female_kkc = female[female['province_of_onset'] == 'ขอนแก่น']
female_kkc_numeric = pd.to_numeric(female_kkc['age'], errors='coerce')
print(female_kkc_numeric.head())
```

```python
sum = 0
N = 0
for a in female_kkc_numeric :
    if a > 0:
        sum += a
        N += 1

print(f'อายุเฉลี่ยของผู้ป่วยหญิงในจังหวัดขอนแก่น {sum/N}')
```

```python
male_kkc = male[male['province_of_onset'] == 'ขอนแก่น']
male_kkc_numeric = pd.to_numeric(male_kkc['age'], errors='coerce')
print(male_kkc_numeric.head())
```

```python
sum = 0
N = 0
for a in male_kkc_numeric :
    if a > 0:
        sum += a
        N += 1

print(f'อายุเฉลี่ยของผู้ป่วยชายในจังหวัดขอนแก่น {sum/N}')
```
